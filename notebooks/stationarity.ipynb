````python
import numpy as np
from statsmodels.tsa.stattools import adfuller, kpss
import statsmodels.api as sm

rng = np.random.default_rng(7)
n = 300
e = rng.normal(size=n)
````

### 1. Four kinds of series

````python
# 1. Four kinds of series
white = e
ar1 = np.zeros(n)
for t in range(1, n):
    ar1[t] = 0.6 * ar1[t - 1] + e[t]
walk = np.cumsum(e)
trend = 0.05 * np.arange(n) + e
````

### 2. ADF test (H0: unit root). Here: constant only and 2 lagged differences

````python
# 2. ADF test (H0: unit root). Here: constant only and 2 lagged differences
def adf_row(name, x, reg="c"):
    stat, p, usedlag, nobs, crit = adfuller(x, maxlag=2, regression=reg, autolag=None)[:5]
    verdict = "stationary" if stat < crit["5%"] else "unit root"
    print(f"{name:14s} tau={stat:7.2f} 5% crit={crit['5%']:6.2f} p={p:.3f} -> {verdict}")

adf_row("white noise", white)
adf_row("AR(1) 0.6", ar1)
adf_row("random walk", walk)
adf_row("trend + noise", trend)                  # fooled: a constant-only test cannot see a trend
adf_row("trend (ct test)", trend, reg="ct")
````

### 3. Differencing a random walk

````python
# 3. Differencing a random walk
adf_row("walk, diff", np.diff(walk))
````

### 4. KPSS has the opposite null (H0: stationary)

````python
# 4. KPSS has the opposite null (H0: stationary)
for name, x in [("AR(1)", ar1), ("random walk", walk)]:
    stat, p, lags, crit = kpss(x, regression="c", nlags=10)
    print(f"KPSS {name:12s} stat={stat:5.2f}  5% crit={crit['5%']:.3f} -> {'reject stationarity' if stat > crit['5%'] else 'stationary'}")
````

### 5. Theory check: moments of a stationary AR(1)

````python
# 5. Theory check: moments of a stationary AR(1)
c, phi, s2 = 2.0, 0.5, 3.0
print("mean", c / (1 - phi), "| variance", round(s2 / (1 - phi ** 2), 3), "| rho(3)", phi ** 3)
````
