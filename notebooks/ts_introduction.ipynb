````python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
````

### 1. Load a real quarterly series and look at its structure

````python
# 1. Load a real quarterly series and look at its structure
df = sm.datasets.macrodata.load_pandas().data
df.index = pd.period_range("1959Q1", periods=len(df), freq="Q")
y = df["realgdp"]                      # real GDP of the USA, billions of dollars
print("observations:", len(y), "| frequency:", y.index.freq, "| from", y.index[0], "to", y.index[-1])
print(y.tail(4).round(1).to_string())
````

### 2. Four benchmark forecasts for the last 8 quarters

````python
# 2. Four benchmark forecasts for the last 8 quarters
train, test = y.iloc[:-8], y.iloc[-8:]
h = len(test)
naive = np.repeat(train.iloc[-1], h)
seasonal_naive = np.array([train.iloc[-4 + (i % 4)] for i in range(h)])
mean_f = np.repeat(train.mean(), h)
drift = train.iloc[-1] + np.arange(1, h + 1) * (train.iloc[-1] - train.iloc[0]) / (len(train) - 1)

def scores(f):
    e = test.values - f
    return round(float(np.sqrt(np.mean(e ** 2))), 1), round(float(np.mean(np.abs(e))), 1)

for name, f in [("naive", naive), ("seasonal naive", seasonal_naive), ("mean", mean_f), ("drift", drift)]:
    print(f"{name:15s} RMSE, MAE = {scores(f)}")
````

### 3. Additive or multiplicative? Compare the spread of changes before and after a log

````python
# 3. Additive or multiplicative? Compare the spread of changes before and after a log
d1 = y.diff().dropna()
d1_log = np.log(y).diff().dropna()
half = len(d1) // 2
print("spread of changes, first half vs second half (raw):", round(d1.iloc[:half].std(), 1), round(d1.iloc[half:].std(), 1))
print("spread of changes, first half vs second half (log):", round(d1_log.iloc[:half].std(), 4), round(d1_log.iloc[half:].std(), 4))
````

### 4. Plot the series and the benchmark forecasts

````python
# 4. Plot the series and the benchmark forecasts
ax = y.iloc[-40:].plot(label="actual")
pd.Series(drift, index=test.index).plot(ax=ax, label="drift")
pd.Series(naive, index=test.index).plot(ax=ax, label="naive")
ax.legend(); plt.show()
````
