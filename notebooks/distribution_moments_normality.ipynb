````python
import numpy as np
from scipy import stats
import statsmodels.api as sm
import matplotlib.pyplot as plt
````

### 1. Moments of a short sample (the numerical problem)

````python
# 1. Moments of a short sample (the numerical problem)
y = np.array([3, 5, 9, 4, 12, 7], dtype=float)
m2 = np.mean((y - y.mean()) ** 2)
m3 = np.mean((y - y.mean()) ** 3)
m4 = np.mean((y - y.mean()) ** 4)
print("mean", round(y.mean(), 3), "| s^2 (n-1)", round(y.var(ddof=1), 3))
print("skewness", round(m3 / m2 ** 1.5, 3), "| kurtosis", round(m4 / m2 ** 2, 3))
print("scipy skew/kurt:", round(stats.skew(y), 3), round(stats.kurtosis(y, fisher=False), 3))
````

### 2. Jarque-Bera by hand and with scipy

````python
# 2. Jarque-Bera by hand and with scipy
rng = np.random.default_rng(1)
samples = {"normal": rng.normal(size=500),
           "log-normal (right skew)": np.exp(0.8 * rng.normal(size=500)),
           "t with 3 df (fat tails)": rng.standard_t(3, size=500)}
for name, x in samples.items():
    S, K, n = stats.skew(x), stats.kurtosis(x, fisher=False), len(x)
    jb = n / 6 * (S ** 2 + (K - 3) ** 2 / 4)
    jb_sp, p = stats.jarque_bera(x)
    print(f"{name:24s} S={S:6.2f} K={K:6.2f} JB={jb:8.2f} (scipy {jb_sp:8.2f}) p={p:.4f}")
````

### 3. A real series: yearly sunspot activity is skewed

````python
# 3. A real series: yearly sunspot activity is skewed
ss = sm.datasets.sunspots.load_pandas().data["SUNACTIVITY"]
print("sunspots: skew", round(stats.skew(ss), 2), "| JB p-value", round(stats.jarque_bera(ss)[1], 6))
print("after sqrt transform: skew", round(stats.skew(np.sqrt(ss)), 2), "| JB p-value", round(stats.jarque_bera(np.sqrt(ss))[1], 4))
````

### 4. Q-Q plots

````python
# 4. Q-Q plots
fig, axes = plt.subplots(1, 2, figsize=(9, 3.5))
sm.qqplot(samples["normal"], line="s", ax=axes[0]); axes[0].set_title("normal")
sm.qqplot(samples["t with 3 df (fat tails)"], line="s", ax=axes[1]); axes[1].set_title("fat tails")
plt.show()
````
